```python
import pandas as pd
import os

DATA_DIR = "data"

tablas = {}
for archivo in os.listdir(DATA_DIR):
    if archivo.endswith(".csv"):
        nombre = archivo.replace("olist_", "").replace("_dataset.csv", "").replace(".csv", "")
        tablas[nombre] = pd.read_csv(os.path.join(DATA_DIR, archivo))

for nombre, df in tablas.items():
    print(f"{nombre:<40} | {df.shape[0]:>8} filas | {df.shape[1]} columnas")
```

```python
tablas["orders"].head()
```

```python
columnas_fecha = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date",
]

for col in columnas_fecha:
    tablas["orders"][col] = pd.to_datetime(tablas["orders"][col])

tablas["orders"]["order_status"].value_counts()
```

```python
mask = (tablas["orders"]["order_status"] == "delivered") & (tablas["orders"]["order_delivered_customer_date"].isna())
tablas["orders"][mask]
```

```python
mask2 = (tablas["orders"]["order_status"] != "delivered") & (tablas["orders"]["order_delivered_customer_date"].notna())
tablas["orders"][mask2][["order_id", "order_status", "order_delivered_customer_date"]]
```

```python
orders = tablas["orders"]
entregados = orders[(orders["order_status"] == "delivered") & (orders["order_delivered_customer_date"].notna())].copy()

entregados["dias_entrega"] = (entregados["order_delivered_customer_date"] - entregados["order_purchase_timestamp"]).dt.days
entregados["dias_prometidos"] = (entregados["order_estimated_delivery_date"] - entregados["order_purchase_timestamp"]).dt.days
entregados["dias_vs_promesa"] = (entregados["order_delivered_customer_date"] - entregados["order_estimated_delivery_date"]).dt.days
entregados["entrega_a_tiempo"] = entregados["dias_vs_promesa"] <= 0

print(f"Pedidos analizados: {len(entregados)}")
print(f"Demora promedio: {entregados['dias_entrega'].mean():.1f} días")
print(f"% entregas a tiempo: {entregados['entrega_a_tiempo'].mean()*100:.1f}%")
entregados["dias_entrega"].describe()
```

```python
print(f"Pedidos con más de 30 días:  {(entregados['dias_entrega'] > 30).sum()}  ({(entregados['dias_entrega'] > 30).mean()*100:.2f}%)")
print(f"Pedidos con más de 60 días:  {(entregados['dias_entrega'] > 60).sum()}  ({(entregados['dias_entrega'] > 60).mean()*100:.2f}%)")
print(f"Pedidos con más de 100 días: {(entregados['dias_entrega'] > 100).sum()}")

entregados.nlargest(10, "dias_entrega")[["order_id", "order_purchase_timestamp", "order_delivered_customer_date", "dias_entrega", "dias_vs_promesa"]]
```

```python
clientes = tablas["customers"][["customer_id", "customer_city", "customer_state"]]
entregados = entregados.merge(clientes, on="customer_id", how="left")

print(f"Filas tras el merge: {len(entregados)}")
print(f"Estados sin asignar: {entregados['customer_state'].isna().sum()}")

resumen_estado = entregados.groupby("customer_state").agg(
    pedidos=("order_id", "count"),
    demora_promedio=("dias_entrega", "mean"),
    pct_a_tiempo=("entrega_a_tiempo", "mean"),
).sort_values("demora_promedio", ascending=False)

resumen_estado.head(10)
```

```python
resumen_estado.tail(5)
```

```python
flete = tablas["order_items"].groupby("order_id").agg(
    items=("order_item_id", "count"),
    valor_productos=("price", "sum"),
    costo_flete=("freight_value", "sum"),
).reset_index()

entregados = entregados.merge(flete, on="order_id", how="left")

print(f"Filas: {len(entregados)}")
print(f"Pedidos sin datos de items: {entregados['items'].isna().sum()}")
entregados[["dias_entrega", "costo_flete", "valor_productos"]].describe()
```

```python
tabla_final = entregados[[
    "order_id", "order_purchase_timestamp", "order_delivered_customer_date",
    "order_estimated_delivery_date", "customer_state", "customer_city",
    "dias_entrega", "dias_prometidos", "dias_vs_promesa", "entrega_a_tiempo",
    "items", "valor_productos", "costo_flete",
]].rename(columns={
    "order_id": "id_pedido",
    "order_purchase_timestamp": "fecha_compra",
    "order_delivered_customer_date": "fecha_entrega",
    "order_estimated_delivery_date": "fecha_prometida",
    "customer_state": "estado",
    "customer_city": "ciudad",
    "items": "cantidad_items",
    "valor_productos": "valor_productos",
    "costo_flete": "costo_flete",
})

# Segmento de criticidad para el dashboard
tabla_final["segmento_entrega"] = pd.cut(
    tabla_final["dias_entrega"],
    bins=[-1, 7, 15, 30, 999],
    labels=["Rápida (0-7)", "Normal (8-15)", "Lenta (16-30)", "Crítica (>30)"],
)

tabla_final.to_csv("data/pedidos_logistica.csv", index=False)
print(f"Exportado: {len(tabla_final)} filas, {len(tabla_final.columns)} columnas")
tabla_final.head()
```
